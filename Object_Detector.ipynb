```python
!pip install -q ultralytics
from google.colab import drive, files
import zipfile
import os

drive.mount('/content/drive')

# Pointing to the correct double-extension path found in your Drive
zip_path = "/content/drive/MyDrive/rickshaw.zip.zip"
if not os.path.exists(zip_path):
    print(f"❌ Could not find: {zip_path}")
    print("Here are the files in your MyDrive directory to help you find the correct name:")
    all_files = os.listdir("/content/drive/MyDrive")
    zip_files = [f for f in all_files if f.endswith('.zip')]
    print("Zip files found:", zip_files if zip_files else "None found in root directory")
else:
    print(f"✅ Found: {zip_path}. Extracting...")
    zipfile.ZipFile(zip_path).extractall("Rick")
    from ultralytics import YOLO
    m = YOLO("yolo11n.pt")
    m.train(data="/content/Rick/data.yaml", epochs=5, batch=32,
            project="/content/drive/MyDrive/yolo_runs", name="rick")
    files.download(str(m.trainer.best))
```
